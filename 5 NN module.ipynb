```python
# create model
import torch
import torch.nn as nn

class model(nn.Module):
  def __init__(self, num_features):
    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(num_features, 3),
        nn.ReLU(),
        nn.Linear(3, 1),
        nn.Sigmoid()
    )



  def forward(self, features):
    z = self.network(features)
    return z
```

```python
# create dataset
features = torch.rand(10, 5)

# create model
model = model(features.shape[1])

# forward pass
model(features)

```

```python
# model weights
print(f"layer1: {model.network[0].weight}  \n")

print(f"layer2: {model.network[2].weight}  \n")
```

```python
# model bias
print(f"layer1: {model.network[0].bias}  \n")

print(f"layer2: {model.network[2].bias}  \n")
```
