```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')

df.head()
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace=True)
```

```python
x = df.iloc[:, 1:]
y = df.iloc[:, 0]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
y_train
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
y_train
```

```python
x_train_tensor = torch.from_numpy(x_train)
x_test_tensor = torch.from_numpy(x_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

# Define model

```python
x.shape[1]
```

```python
import torch
import torch.nn as nn

class MySimpleNN(nn.Module):
  def __init__(self, num_features):
    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(num_features, 1),
        nn.Sigmoid()
    )

  def forward(self, feature):
    z =  self.network(feature)
    return z
```

```python
learning_rate = 0.1
epochs = 20
batch_size = 32
```

# training pipeline

```python
loss_function = nn.BCELoss()
```

```python
# create pipeline

model = MySimpleNN(x_train_tensor.shape[1])

#Optimizers
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

# Convert tensors to float32 and reshape targets to match [N, 1]
x_train_tensor = x_train_tensor.float()
y_train_tensor = y_train_tensor.float()

#define loop
for epoch in range(epochs):
  for start_idx in range(0, x_train_tensor.size(0), batch_size):
    end_idx = start_idx + batch_size


    #creating batch
    x_batch = x_train_tensor[start_idx:end_idx]
    y_batch = y_train_tensor[start_idx:end_idx]


    # forward pass
    y_pred = model(x_batch)

    # loss calculation
    loss = loss_function(y_pred, y_batch.view(-1, 1))


    # zero gradients
    optimizer.zero_grad()

    # backward pass
    loss.backward()

    # update weights
    optimizer.step()


  print(f'epoch:{epoch+1} loss={loss.item()}')
```
