```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')

df.head()
```

```python
df.shape
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace=True)
```

```python
df.shape
```

```python
x = df.iloc[:, 1:]
y = df.iloc[:, 0]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
y_train
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_train
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
y_train
```

```python
x_train_tensor = torch.from_numpy(x_train)
x_test_tensor = torch.from_numpy(x_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

```python
x_train_tensor
```

```python
x_train_tensor.shape
```

# Define model

```python
x.shape[1]
```

```python
import torch
import torch.nn as nn

class MySimpleNN(nn.Module):
  def __init__(self, num_features):
    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(num_features, 1),
        nn.Sigmoid()
    )

  def forward(self, feature):
    z =  self.network(feature)
    return z

  # def lossf(self, y_pred, y):
  #   eps = 1e-7
  #   y_pred = torch.clamp(y_pred, eps, 1 - eps)
  #   loss = -(y * torch.log(y_pred) + (1 - y) * torch.log(1 - y_pred)).mean()
  #   return loss
```

```python
learning_rate = 0.1
epochs = 20
```

# training pipeline

```python
loss_function = nn.BCELoss()
```

```python
# create pipeline

model = MySimpleNN(x_train_tensor.shape[1])

#Optimizers
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

# Convert tensors to float32 and reshape targets to match [N, 1]
x_train_f32 = x_train_tensor.float()
y_train_f32 = y_train_tensor.float()

#define loop
for epoch in range(epochs):
  # forward pass
  y_pred = model(x_train_f32)

  # loss calculation
  loss = loss_function(y_pred, y_train_f32.view(-1, 1))
  # print(loss)

  # zero gradients
  optimizer.zero_grad()

  # backward pass
  loss.backward()

  # update weights
  optimizer.step()


  print(f'epoch:{epoch+1} loss={loss.item()}')
```

# Evalution

```python
# model evalution

with torch.no_grad():
  # Convert the test features and targets to Float32 to match the model's dtype
  x_test_f32 = x_test_tensor.float()
  y_test_f32 = y_test_tensor.float()

  y_pred = model(x_test_f32)
  y_pred = (y_pred > 0.5).float()

  accuracy = (y_pred == y_test_f32).float().mean()
  print(f'accuracy:{accuracy.item()}')
```
