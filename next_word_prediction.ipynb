```python
# tokenize
tokens = word_tokenize(text=document.lower())
```

```python
# build vocab

vocab = {'<unk>':0}

for token in Counter(tokens).keys():
  if token not in vocab:
    vocab[token] = len(vocab)


print(vocab)
```

```python
# extract sentences from data

input_sentences = document.split('\n')
```

```python
def text_to_sequence(sentences, vocab):
  numerical_sentence = []
  for token in sentences:
      if token in vocab:
        numerical_sentence.append(vocab[token])
      else:
        numerical_sentence.append(vocab['<unk>'])
  return numerical_sentence
```

```python
input_numerial_sentences = []

for sentence in input_sentences:
  input_numerial_sentences.append(text_to_sequence(sentences=word_tokenize(text=sentence.lower()), vocab=vocab))
```

```python
class CustomDataset(Dataset):
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def __len__(self):
    return self.x.shape[1]

  def __getitem__(self, index):
    # print('statement: ' ,str(self.x[index]))
    # print('target: ' + str(self.y[index]))
    return self.x[index], self.y[index]
```

# model building


```python
class LSTMModel(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.embedding = nn.Embedding(num_embeddings=vocab_size, embedding_dim=100)
        self.lstm = nn.LSTM(input_size=100, hidden_size=150, batch_first=True)
        self.fc = nn.Linear(in_features=150, out_features=vocab_size)



    def forward(self, x):
        x = self.embedding(x)
        (hiddenState, (final_hiddenState, final_CellState)) = self.lstm(x)
        final_hiddenState = final_hiddenState.squeeze(0)
        final_output = self.fc(final_hiddenState)

        return final_output
```

```python
x = nn.Embedding(num_embeddings=len(vocab), embedding_dim=100)
y = nn.LSTM(input_size=100, hidden_size=150, batch_first=True)
z = nn.Linear(in_features=150, out_features=len(vocab))
```

```python
# building model
model = LSTMModel(vocab_size=len(vocab))
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
print(torch.cuda.is_available())
```

```python
model.to(device)
```

```python
epochs = 50
learning_rate = 0.001

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)
```

```python
# training loop
for epoch in range(epochs):
    total_loss = 0
    for statement, target in dataloader:
        statement = statement.to(device)
        target = target.to(device)

        # clearing all gradiant
        optimizer.zero_grad()

        #forward prop
        output = model(statement)

        #loss
        loss = criterion(output, target)

        #backprop
        loss.backward()

        #update
        optimizer.step()

        total_loss += loss.item()

    print(f'Epoch: {epoch+1}, Loss: {total_loss/len(dataloader)}')
```

```python
#prediction
def prediction(model, vocab, text):

    #tokenize
    tokenize_text = word_tokenize(text=text.lower())
    # print( tokenize_text)

    #text -> numerical indices
    numerical_text = text_to_sequence(sentences=tokenize_text, vocab=vocab)
    # print(numerical_text)

    # padding
    padd_numerical_text = [0]*(max_len-len(numerical_text)) + numerical_text
    padd_numerical_text = torch.tensor(padd_numerical_text, dtype=torch.long).unsqueeze(0)
    # print(padd_numerical_text.shape)

    # send to model
    output = model(padd_numerical_text.to(device))

    # predicted indecx
    prob, index = torch.max(output, dim=1)
    # print(index)
    next_word = (list(vocab.keys())[index])

    # merge with text
    text+" "+next_word
    return text+" "+next_word
```
