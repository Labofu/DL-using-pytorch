```python
# create model
import torch
import torch.nn as nn

class model(nn.Module):
  def __init__(self, num_features):
    super().__init__()
    self.linear1 = nn.Linear(num_features, 3)
    self.relu = nn.ReLU()
    self.linear2 = nn.Linear(3, 1)
    self.sigmoid = nn.Sigmoid()



  def forward(self, features):
    z = self.linear1(features)
    z = self.relu(z)
    z = self.linear2(z)
    z = self.sigmoid(z)
    return z
```

```python
# create dataset
features = torch.rand(10, 5)

# create model
model = model(features.shape[1])

# forward pass
model(features)

```

```python
# model weights
print(f"layer1: {model.linear1.weight}  \n")

print(f"layer2: {model.linear2.weight}  \n")



```

```python
# model bias
print(f"layer1: {model.linear1.bias}  \n")

print(f"layer2: {model.linear2.bias}  \n")
```
